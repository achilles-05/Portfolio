```python
# ============================================================
# GRAPHITE EXPORT VALUE FORECAST TILL DEC 2027
# (Monthly data, Gold-style LSTM)
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from keras import Model
from keras.layers import Input, Dense, Dropout, LSTM
from google.colab import files

# ------------------------------------------------------------
# Upload CSV
# ------------------------------------------------------------
uploaded = files.upload()
file_name = list(uploaded.keys())[0]

# ------------------------------------------------------------
# Load & prepare data
# ------------------------------------------------------------
df = pd.read_csv(file_name)

# Convert Year to datetime (CRITICAL FIX)
df["Year"] = pd.to_datetime(df["Year"], format="%Y-%m")

df = df.sort_values("Year").reset_index(drop=True)

df["Export"] = df["Export"].astype("float64")

dates = df["Year"].values
export_values = df["Export"].values.reshape(-1,1)

last_date = df["Year"].iloc[-1]
print("Last date in data:", last_date)

# ------------------------------------------------------------
# Scale data
# ------------------------------------------------------------
scaler = MinMaxScaler()
scaled_export = scaler.fit_transform(export_values)

# ------------------------------------------------------------
# Create sequences
# ------------------------------------------------------------
window_size = 12

X, y = [], []
for i in range(window_size, len(scaled_export)):
    X.append(scaled_export[i-window_size:i, 0])
    y.append(scaled_export[i, 0])

X = np.array(X).reshape(-1, window_size, 1)
y = np.array(y).reshape(-1,1)

print("Training samples:", X.shape)

# ------------------------------------------------------------
# Define model (UNCHANGED)
# ------------------------------------------------------------
def define_model():
    input1 = Input(shape=(window_size, 1))
    x = LSTM(64, return_sequences=True)(input1)
    x = Dropout(0.2)(x)
    x = LSTM(64, return_sequences=True)(x)
    x = Dropout(0.2)(x)
    x = LSTM(64)(x)
    x = Dropout(0.2)(x)
    x = Dense(32, activation="softmax")(x)
    output = Dense(1)(x)

    model = Model(inputs=input1, outputs=output)
    model.compile(loss="mean_squared_error", optimizer="Nadam")
    return model

model = define_model()
model.summary()

# ------------------------------------------------------------
# Train on ALL data
# ------------------------------------------------------------
model.fit(
    X, y,
    epochs=150,
    batch_size=32,
    verbose=1
)

# ------------------------------------------------------------
# CREATE FUTURE MONTHS TILL DEC 2027
# ------------------------------------------------------------
future_dates = pd.date_range(
    start=last_date + pd.DateOffset(months=1),
    end="2027-12-01",
    freq="MS"
)

print("Forecasting months:", len(future_dates))

# ------------------------------------------------------------
# Recursive forecasting
# ------------------------------------------------------------
last_window = scaled_export[-window_size:].reshape(1, window_size, 1)

future_predictions_scaled = []

for _ in range(len(future_dates)):
    next_pred = model.predict(last_window, verbose=0)[0,0]
    future_predictions_scaled.append(next_pred)

    last_window = np.append(
        last_window[:,1:,:],
        [[[next_pred]]],
        axis=1
    )

# Inverse scale
future_predictions = scaler.inverse_transform(
    np.array(future_predictions_scaled).reshape(-1,1)
)

# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------
plt.figure(figsize=(15,6), dpi=150)

plt.plot(df["Year"], df["Export"], color="black", lw=2, label="Historical Data")
plt.plot(future_dates, future_predictions, color="red", lw=2, marker="o", label="Forecast (till Dec 2027)")

plt.axvline(x=last_date, color="gray", linestyle="--", label="Forecast Start")

plt.title("Graphite Export VALUE Forecast till Dec 2027 (LSTM)")
plt.xlabel("Date")
plt.ylabel("Export Value")
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# Forecast table
# ------------------------------------------------------------
forecast_df = pd.DataFrame({
    "Date": future_dates,
    "Predicted_Export": future_predictions.flatten()
})

print("\nForecast till Dec 2027:")
print(forecast_df)
```
